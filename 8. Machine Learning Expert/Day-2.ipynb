# Gradient Descent in Machine Learning

## 1. Linear Regression

Linear Regression is a supervised machine learning algorithm used to predict continuous values based on input features.

**Formula:**


ŷ = mx + c


Where:
- `m` = Slope (Weight)
- `c` = Intercept (Bias)
- `x` = Input feature
- `ŷ` = Predicted output

## 2. What is Gradient?

Gradient is a mathematical vector that indicates the direction and rate of the steepest increase of a function.

In Machine Learning, Gradient helps us understand how changes in model parameters affect the loss function.

**Formula:**

```math
∇J(θ)
```

Where:
- `∇` = Gradient operator
- `J` = Cost function
- `θ` = Model parameters

**In simple terms:** Gradient tells us the direction in which the error increases and how rapidly it increases.

## 3. What is Descent?

Descent means moving towards a lower value of a function.

In Machine Learning, our objective is to minimize the cost function to reduce the prediction error.

**In simple terms:** Descent means moving towards the minimum error.

## 4. What is Gradient Descent?

Gradient Descent is an iterative optimization algorithm used in Machine Learning to minimize the cost function by repeatedly updating model parameters in the opposite direction of the gradient.

**Main Formula:**

```math
θ_new = θ_old - α ∇J(θ)
```

Where:
- `θ` = Model parameters
- `α` = Learning rate
- `∇J(θ)` = Gradient of the cost function
- `J(θ)` = Cost function

## 5. Cost Function (Mean Squared Error)

Mean Squared Error (MSE) measures the average squared difference between actual and predicted values.

**Formula:**

```math
J(m,c) = (1/n) Σᵢ₌₁ⁿ (yᵢ - ŷᵢ)²
```

Where:

```math
ŷᵢ = mxᵢ + c
```

- `n` = Number of data points
- `yᵢ` = Actual value
- `ŷᵢ` = Predicted value

**Objective:** Minimize the cost function.

## 6. Partial Derivatives of m and c

In Linear Regression, we update two parameters:

- `m` = Slope (Weight)
- `c` = Intercept (Bias)

### A. Derivative with respect to m

```math
∂J/∂m = (-2/n) Σᵢ₌₁ⁿ xᵢ(yᵢ - ŷᵢ)
```

### B. Derivative with respect to c

```math
∂J/∂c = (-2/n) Σᵢ₌₁ⁿ (yᵢ - ŷᵢ)
```

## 7. Gradient Descent Update Formulas

### A. Update formula for m (Slope)

```math
m_new = m_old - α(∂J/∂m)
```

### B. Update formula for c (Intercept)

```math
c_new = c_old - α(∂J/∂c)
```

### Simplified formulas

```math
m_new = m_old + (2α/n) Σᵢ₌₁ⁿ xᵢ(yᵢ - ŷᵢ)
```

```math
c_new = c_old + (2α/n) Σᵢ₌₁ⁿ (yᵢ - ŷᵢ)
```

## 8. Steps Involved in Gradient Descent

1. Initialize parameters `m` and `c`.
2. Calculate predicted values using `ŷ = mx + c`.
3. Calculate the cost function (MSE).
4. Calculate partial derivatives for `m` and `c`.
5. Update `m` and `c` using Gradient Descent.
6. Repeat until the cost function converges or a stopping condition is reached.

## 9. Types of Gradient Descent

| Type | Explanation |
|---|---|
| Batch Gradient Descent | Uses the entire training dataset for each update. |
| Stochastic Gradient Descent (SGD) | Uses one randomly selected training example for each update. |
| Mini-Batch Gradient Descent | Uses a small group of training examples for each update. |

## 10. Learning Rate (α)

Learning Rate determines the step size taken by Gradient Descent while updating model parameters.

- **Small Learning Rate:** Training becomes slow.
- **Large Learning Rate:** May overshoot the minimum and fail to converge.
- **Appropriate Learning Rate:** Helps achieve efficient convergence.

## 11. Difference Between Gradient and Descent

| Gradient | Descent |
|---|---|
| Indicates the direction and rate of steepest increase. | Moves towards a lower function value. |
| Calculated using derivatives. | Uses the negative gradient to move downhill. |
| Helps determine parameter updates. | Helps minimize the cost function. |



# Regression Evaluation Metrics in Machine Learning

## 1. MSE (Mean Squared Error)

**Definition:** Mean Squared Error measures the average of the squared differences between actual values and predicted values.

**Formula:**

```text
MSE = (1/n) * Σ (yᵢ - ŷᵢ)²
```

**Key Points:**
- Squares the prediction errors.
- Penalizes large errors more heavily.
- Lower MSE indicates better model performance.
- MSE is always non-negative.

## 2. MAE (Mean Absolute Error)

**Definition:** Mean Absolute Error measures the average absolute difference between actual values and predicted values.

**Formula:**

```text
MAE = (1/n) * Σ |yᵢ - ŷᵢ|
```

**Key Points:**
- Uses absolute errors.
- Less sensitive to outliers than MSE.
- Easy to interpret.
- Lower MAE indicates better model performance.

## 3. RMSE (Root Mean Squared Error)

**Definition:** Root Mean Squared Error is the square root of Mean Squared Error.

**Formula:**

```text
RMSE = √[(1/n) * Σ (yᵢ - ŷᵢ)²]
```

**Key Points:**
- Square root of MSE.
- Expressed in the same unit as the target variable.
- Penalizes large errors.
- Lower RMSE indicates better model performance.

## 4. R² Score (Coefficient of Determination)

**Definition:** R² Score measures how much of the variation in the dependent variable is explained by the regression model.

**Formula:**

```text
R² = 1 - (SSE / SST)
```

**Key Points:**
- Maximum value is 1.
- R² = 1 indicates a perfect fit.
- R² = 0 means the model performs like the mean-prediction baseline.
- R² can be negative.
- Higher R² generally indicates better explanatory performance.

## 5. SSR (Regression Sum of Squares)

**Definition:** SSR measures the variation in predicted values around the mean of actual values. It represents the variation explained by the regression model.

**Formula:**

```text
SSR = Σ (ŷᵢ - ȳ)²
```

Where:
- `ŷᵢ` = Predicted value
- `ȳ` = Mean of actual values

## 6. SSE (Error Sum of Squares)

**Definition:** SSE measures the variation between actual values and predicted values. It represents the unexplained variation.

**Formula:**

```text
SSE = Σ (yᵢ - ŷᵢ)²
```

## 7. SST (Total Sum of Squares)

**Definition:** SST measures the total variation of actual values around their mean.

**Formula:**

```text
SST = Σ (yᵢ - ȳ)²
```

Where:
- `yᵢ` = Actual value
- `ȳ` = Mean of actual values

## 8. Relationship Between SSR, SSE and SST

For standard Ordinary Least Squares (OLS) regression with an intercept:

```text
SST = SSR + SSE
```

Where:
- `SST` = Total Sum of Squares
- `SSR` = Regression Sum of Squares (Explained Variation)
- `SSE` = Error Sum of Squares (Unexplained Variation)

**Note:** Some textbooks use SSR to mean Residual Sum of Squares. Always check the notation used in the question.

## 9. Comparison Table

| Metric | Full Form | Measures |
|---|---|---|
| MSE | Mean Squared Error | Average squared error |
| MAE | Mean Absolute Error | Average absolute error |
| RMSE | Root Mean Squared Error | Root of average squared error |
| R² | Coefficient of Determination | Explained variation |
| SSR | Regression Sum of Squares | Explained variation |
| SSE | Error Sum of Squares | Unexplained variation |
| SST | Total Sum of Squares | Total variation |

## 10. Important Formulas for GATE DA

```text
MSE  = (1/n) * Σ (yᵢ - ŷᵢ)²

MAE  = (1/n) * Σ |yᵢ - ŷᵢ|

RMSE = √MSE

R²   = 1 - (SSE / SST)

SSR  = Σ (ŷᵢ - ȳ)²

SSE  = Σ (yᵢ - ŷᵢ)²

SST  = Σ (yᵢ - ȳ)²

SST  = SSR + SSE
```

## 11. Important Points to Remember

1. Lower MSE, MAE and RMSE indicate better predictive performance.
2. Higher R² generally indicates better explanatory performance.
3. MSE is more sensitive to outliers than MAE.
4. RMSE and MAE are expressed in the original target variable's units.
5. R² can be negative.
6. SST = SSR + SSE for standard OLS regression with an intercept.
7. Perfect prediction gives MSE = MAE = RMSE = 0 and R² = 1, provided SST is nonzero.
