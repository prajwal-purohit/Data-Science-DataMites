### Sigmoid Function

$$
y = \frac{1}{1 + e^{-z}}
$$

Where,

$$
z = m_0 + m_1x_1 + m_2x_2 + \cdots + m_nx_n
$$

$$
e = Euler's Number = 2.71818
$$

**Output Range:**

$$
0 < y < 1
$$

### Evaluation Metrices for Classification

1. True Positive (TP)
2. False Negative (FN)
3. False Positive (FP)
4. True Negative (TN)

### 1. Accuracy
$$
Accuracy = \frac{TP + TN}{TP + TN + FP + FN}
$$

### 2. Precision
$$
Precision = \frac{TP}{TP + FP}
$$

### 3. Recall
$$
Recall = \frac{TP}{TP + FN}
$$

### 4. F1 Score
$$
F1\ Score = 2 \times \frac{Precision \times Recall}{Precision + Recall}
$$

### Terms:
- TP = True Positive
- TN = True Negative
- FP = False Positive
- FN = False Negative
