# MACHINE LEARNING

### Types of ML
1. **Supervised Learning**: Labelled Data (Features and Targets)
2. **Unsupervised Learning**: Unlabelled Data (Only Features)
3. **Semi-Supervised Learning**: Semi-labelled (20% of Unsupervised Learning and 80% of Supervised Learning)
4. **Reinforcement Learning**: Reaward and Penalty

### Types of Supervised Learning

1. **Regression**- Use only if Datatype is Numerical Continuous 
2. **Classification**- Use Datatype is Numerical Discrete and Categorical

### Linear Regression

- **Goal**: The Goal is to find Straight Line Relation between features and target.

- **Simple LR**: 1 feature and 1 target

x   y
1   10
2   20
3   30
4   40
5   50

here y = 10x from straight line equation: y = mx + c, where m is slope and c is interception
