### Random Number Generation
- randn: Generates random number of given shape in a SND form.
- rand: Generates random number in the interval 0, 1 (0 is included and 1 is excluded).
- randint: Generates random integers with given start, stop, and shape.

```python
rand1 = np.random.rand(2, 3)
rand1
```

```python
rand2 = np.random.rand(2, 3, 1)
rand2
```

```python
rand3 = np.random.randn(3, 3)
rand3
```

```python
rand4 = np.random.randn(3, 2, 1)
rand4
```

```python
rand5 = np.random.randint(15, 27, 4)
rand5
```

```python
rand6 = np.random.randint(15, 27, (3, 2, 2))
rand6
```

### Performance Comparision

```python
import numpy as np
import time
```

```python
a = list(range(1000000))
b = list(range(1000000))
```

```python
start = time.time()
c = [a[i] + b[i] for i in range(len(a))]
print("Loop Time : ", time.time() - start)
```

```python
a_np = np.array(a)
b_np = np.array(b)
start = time.time()
c_np = a_np + b_np
print("Vectorized Time : ", time.time() - start)
```
