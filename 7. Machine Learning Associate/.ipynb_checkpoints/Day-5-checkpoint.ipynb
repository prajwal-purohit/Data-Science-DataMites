### Changing of Datatype
- astype
- downcast

```python
df.info()
```

```python
df['Survived'] = df.Survived.astype('float16')
```

```python
df.info()
```

```python
df['Pclass'] = df.Pclass.astype('float64')
```

```python
df.info()
```

```python
df
```

```python
df['Ticket'] = df.Ticket.astype('int8')
df.info()
```

```python
df.info()
```

```python
df['PassengerId'] = pd.to_numeric(df['PassengerId'], downcast = 'integer')
df.info()
```

```python
df
```

```python
df['SibSp'] = pd.to_numeric(df['SibSp'], downcast = 'float')
df.info()
```
