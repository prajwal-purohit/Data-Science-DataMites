### Data Visualization

Matplotlib

#### *Line Plot*

```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 15, 13, 18, 16]
```

```python
plt.plot(x, y)
plt.show()
```

### Line Plot with Customized Features

```python
x = [1, 2, 3, 4, 5]
y = [10, 15, 13, 18, 16]
plt.plot(x, y, marker='o', linestyle='--', color='r')
plt.title('Line Plot')
plt.xlabel('x - axis')
plt.ylabel('y - axis')
plt.grid(True)
plt.show()
```

```python
x = [1, 2, 3, 4]
y = [10, 20, 25, 30]
z = [15, 18, 22, 27]
plt.plot(x, y, label = 'Product A')
plt.plot(x, z, label = 'Product B')
plt.legend(loc='upper left')
plt.title('Product Sales Comparision')
plt.xlabel('Days')
plt.ylabel('Units Sold')
plt.show()
```

```python
categories = ['A', 'B', 'C', 'D']
values = [23, 45, 12, 36]

plt.bar(categories, values, color='pink')
plt.title('Bar Plot')
plt.xlabel('Category')
plt.ylabel('Values')
plt.show()
```

### Scatter Plot

```python
x = [5, 7, 8, 7, 2, 17, 2, 9, 4, 1]
y = [99, 86, 87, 86, 100, 86, 103, 87, 94, 78]

plt.scatter(x, y, color='red')
plt.title('Scatter Plot')
plt.xlabel('x values')
plt.ylabel('y values')
plt.show()
```

### Seaborn

```python
import numpy as np
import pandas as pd
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
sns.get_dataset_names()
```

```python
df = sns.load_dataset('iris')
df.head(10)
```

```python
df.species.unique()
```

```python
df.species.value_counts()
```

### Line Plot

```python
sns.lineplot(data = df, x='sepal_length', y='sepal_width', ci = None)
plt.show()
```

### Confidence Interval (CI)
- 

```python
sns.lineplot(data = df, hue = 'species', x='sepal_length', y='sepal_width', ci = None)
plt.show()
```

### Scatter Plot

```python
sns.scatterplot(x = 'sepal_length', y = 'sepal_width', data = df, hue = 'species')
plt.title('Scatter plot')
plt.show()
```

### Bar Plot

```python
sns.barplot(x = 'species' , y = 'sepal_length', data = df, hue = 'species')
plt.title('Bar plot')
plt.show()
```

### Count Plot

```python
df.species.value_counts()
```

```python
sns.countplot(x = 'species', data = df)
plt.show()
```

```python
df
```

### Box Plot
- To Find Outliers

```python
sns.boxplot(x = 'sepal_width', data = df)
plt.show()
```

```python
sns.boxplot(x = 'species', y = 'sepal_width', data = df)
plt.show()
```

### Violin Plot

```python
sns.violinplot(x = 'species', y = 'sepal_width', data = df)
plt.show()
```

### Histplot

```python
sns.histplot(x = 'petal_width', data = df)
plt.show()
```

```python
sns.histplot(x = 'petal_width', data = df, hue = 'species', kde = True)
plt.show()
```

### Distplot

```python
sns.distplot(df['petal_width'], color = 'g')
plt.show()
```
