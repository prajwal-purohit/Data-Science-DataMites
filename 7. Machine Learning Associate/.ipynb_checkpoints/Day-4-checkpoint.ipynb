### Boolean Masks
- We Select rows that meet logical conditions.

```python
df.Age > 28
```

```python
df[df['Age'] > 28]
```

```python
df.loc[(df['Gender'] == 'female') & (df['Age']< 38)]
```

```python
df.loc[(df['Gender'] == 'female') | (df['Age']< 38)]
```

### Missing Values/ Null Values
- For memory management
- NaN---> also missing value
- Dropping
- Imputing

```python
df
```

### When to Drop?
- if % of null value is > 70%

### When to Fill?
- if % of null value is < 30%

### Filling:
- If Numerical Continuous and Normally Distributed ---> Mean
- If Numerical Continuous and not Normally Distributed ---> Median
- If Numerical Discrete or Categorical ---> Mode

```python
df.dropna(axis = 1) # Dropping the Columns which are having the null value
```

```python
df.dropna(axis = 0) # Dropping the Rows which are having the null value
```
