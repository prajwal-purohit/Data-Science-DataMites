### Numpy Arrays
* Numpy arrays essentially of two types: vectors ands matrices.
* Vectors are strictly 1-D arrays
* Matrices are 2-D arrays
* Note: A matrix can still have only one row or one column
* Arrays are heterogeneous
* Arrays are immutable

### Built-in Array Generation :
* Zeroes: generates an array of given shape with zeroes as elements
* Ones: generates an array of given shape with ones as elements
* linspace: generates an array of mentioned element between given start and stop point with equidistantvalues
* eye: generates a 2D array of identity matrix og given dimension
