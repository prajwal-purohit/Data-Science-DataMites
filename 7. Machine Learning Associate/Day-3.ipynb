### Series

```python
import pandas as pd
```

```python
l1 = ['abc', 'A', 19, 'Male']
l1
```

```python
s1 = pd.Series(l1)
s1
```

```python
l2 = ['abc', 'A', 19, 'Male'], ['xyz', 'B', 19, 'Female']
l2
```

```python
s2 = pd.Series(l2)
s2
```

```python
s3 = pd.Series([10, 100, 1000, 10000], index =['I', 'II', 'III', 'IV'])
s3
```

```python
dict1 = {'Laptop':'Dell', 'AC':'Inverter', 'Projector': 'BenQ', 'Fan':'Crompton'}
dict1
```

```python
s4 = pd.Series(dict1)
s4
```

```python
df1 = pd.DataFrame({'Name': ['abc', 'xyz', 'lmn'], 'Age': [19, 18,20], 'Grade': ['A', 'A', 'B']})
df1
```

```python
a5 = [['abc', 'A', 19, 'Male'], ['xyz', 'B', 19, 'Female', 12345], ['lmn', '20', 'Female'], ['pqr', 'A+', 17, 'Male'], ['def', 'B+', 21, 'Male']]
df3 = pd.DataFrame(a5, columns=['Name', 'Grade', 'Age', 'Gender', 'Dummy'])
df3
```

NaN(No any Number), None, Null ---> all are empty

```python
a6 = [['abc', 'A', 49, 'Male'], ['xyz', 'B', 19, 'Female'], ['lmn','B', 20, 'Female'], ['pqr', 'A+', 17, 'Male'], ['def', 'B+', 21, 'Male']]
df2 = pd.DataFrame(a6, columns=['Name', 'Grade', 'Age', 'Gender'])
df2
```

```python
df2.rename({'Age': 'Age_in_years'}, axis=1, inplace=True)
df2
```
