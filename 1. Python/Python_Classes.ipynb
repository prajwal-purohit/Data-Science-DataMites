### SLICING
Variable [Starting Index Number : Ending Index Number Plus One Extra : Step Size]

```python
a = "Aeroplane"

# Forward Slicing
# a[0] ---> A
# a[1] ---> e
# a[2] ---> r
# a[3] ---> o
# a[4] ---> p
# a[5] ---> l
# a[6] ---> a
# a[7] ---> n
# a[8] ---> e

# Backward Slicing
# a[-1] ---> e
# a[-2] ---> n
# a[-3] ---> a
# a[-4] ---> l
# a[-5] ---> p
# a[-6] ---> o
# a[-7] ---> r
# a[-8] ---> e
# a[-9] ---> A
```

```python
# From "Hello World" Extract "Wor"
print(b[6:9:1])

# From "Hello World" Extract "Hello"
print(b[0:5:1])
# Same without Zero
print(b[:5:1])

# From "Hello World" Extract "World"
print(b[6:11:1])
# Same without Last Number
print(b[-5::1])
```

```python
a = "Aeroplane"
# From Aeroplane Extract "rop"
print(a[2:5:1])

# From Aeroplane Extract "Aero"
print(a[:4:1])

# From Aeroplane Extract "plane"
print(a[4:9:1])
# Same without Last Index
print(a[4::1])
```

```python
# Reverse Slicing

# From "Hello World" Extract "roW"
print(b[-3:-6:-1])

# From "Aeroplane" Extract "pore"
print(a[-5:-9:-1])

# From "Hello World" Extract "roW"
print(b[-3:-6:-1])
```

```python
# Reverse the String
print(a[-1:-10:-1])

# Without Extreme Index
print(a[-1::-1])
```

```python
# Example
word = "photosynthesis"

# Get the "synthesis"
print(word[5::1])

# Get the "photo"
print(word[:5:1])

# Get the "syn"
print(word[5:8:1])

# Get the "thes"
print(word[8:-2:1])

# Get the "is"
print(word[-2::1])

# Reverse the "photosyntheis"
print(word[-1::-1])
# Also
print(word[::-1])

# Print the Statement using Slicing
print(word[::1])
```

```python
sp = "TENET"
# Get "NET" from "TENET"
print(sp[2::])
print(sp[-3::-1])

# Get "TEN" from "TENET"
print(sp[:3:])
print(sp[:-4:-1])

# Get "ENE" from "TENET"
print(sp[1:4:])
print(sp[-2:-5:-1])

# Reverse "TENET"
print(sp[::-1])

# Print using Slicing
print(sp[::])
```
