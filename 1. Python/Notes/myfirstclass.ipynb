```python
## Integer Type Data
import sys
a = 1
print(type(a))
print(sys.getsizeof(a))
print(isinstance(a,int))
```

```python
## Float
b = 6.45
print(type(b))
print(sys.getsizeof(b))
print(isinstance(b,float))
print(isinstance(b,int))
```

```python
## String
d = "aeroplane"
e = 'apple'
print(type(d))
print(sys.getsizeof(d))
print(isinstance(d,str))
```

```python
## Boolean
x = True # 1
y = False # 0
print(type(x),type(y))
print(sys.getsizeof(x),sys.getsizeof(y))
print(isinstance(x,bool))
print(isinstance(x,int))
```

```python
a = "Aeroplane"
print(a[3])
## Get the last character
print(a[-1])
b = "Hello World"
# get the space out from Hello world
print(b[5])
# Get W out Hello world
print(b[6])
# get 'rop' out of Aeroplane
print(a[2:5:1])
# get 'Aero' from Aeroplane
print(a[:4:1])
print(a[0:4:1])
# get "Plane" out of Aeroplane
print(a[4:9:1])
print(a[4::1])
# get 'wor' from Hello World
print(b[6:9:1])
```

```python
## Reverse slicing
# Get the 'row' from hello world
print(b[-3:-6:-1])
# get 'pore' from Aeroplane
print(a[-5:-9:-1])
# Reverse the string "Aeroplane"
print(a[-1:-10:-1])
print(a[::-1])
```
