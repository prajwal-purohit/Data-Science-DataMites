# Pypark RDD Basics

RDD is a distributed collection of data, transformation define  what we want to do, while actions trigger. Spark to execute the computation and return/write the result.

## Create Spark Session

```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName('RDD Basics').master("local[*]").getOrCreate()
sc = spark.sparkContext # Spark gives us the connection we need to work with RDDs.
```

## Create an RDD

```python
numbers = [10, 20, 30, 40, 50]

rdd = sc.parallelize(numbers) # Converting connection into RDD
rdd.collect()
```

## Number of Partitions

```python
rdd.getNumPartitions() # RDD Partitions
```

```python
square = rdd.map(lambda x: x*x)
square.collect()
```

```python
a = rdd.filter(lambda x: x % 20 == 0)
a.collect()
```

## Actions

```python
print("Count : ", rdd.count())
print("First : ", rdd.first())
print("Take : ", rdd.take(3))
```

```python
words = sc.parallelize(['Big', 'Data', 'Big', 'Spark', 'Data'])
words.collect()
```

```python
word_count = (words
              .map(lambda x: (x, 1))
              .reduceByKey(lambda a, b: a + b))
word_count.collect()
```
