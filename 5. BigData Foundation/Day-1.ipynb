Starting the Spark Session



```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName('DataFrame Basics').master('local[*]')
```

**Create a DataFrame**

```python
data = [('Ravi', 22, 'Python'),
      ('Priya', 23, 'SQL'),
      ('Rahul', 21, 'PowerBI'),
      ('Anu', 24, 'Python')]
```

Create a Schema

```python
# Get or create the Spark session
spark = SparkSession.builder.appName('DataFrame Basics').master('local[*]').getOrCreate()

columns = ['Name', 'Age', 'Language']
df = spark.createDataFrame(data, columns)
df.show()
```

**Display a Schema**

```python
df.printSchema()
```

Select Schema

```python
df.select('Name', 'Language').show()
```

Group By

```python
df.groupBy('Language').count().show()
```

Filter Data

```python
df.filter(df.Age == 22).show()
```

Group By

```python
df.groupBy('Language').avg('Age').show()
```

```python
from pyspark.sql.functions import *
```

```python
df.groupBy('Language').agg(avg('Age'), count('Name')).show()
```

Stop Spark

```python
spark.stop()
```
