### Mean, Mode, Median, Variance, Standdard Deviation, Percentile :

```python
import numpy as np
train_a = [4,5,5,5,6]
train_b = [7,3,5,9,1]
train_c = [7,7,7,7,7]
print(np.mean(train_a))
print(np.mean(train_b))
print(np.mean(train_c))

train_a_mean = (4 + 5 + 5 + 5 + 6)/5
print(f"Mean of Train A is : {train_a_mean}")
train_a_var = ((4-5)**2 + (5-5)**2 + (5-5)**2 + (5-5)**2 + (6-5)**2)/5
print(f"Variance of Train A is : {train_a_var}")
print(f"Variance of Train A is : {np.var(train_a)}")
print(f"Standard Variation of Train A is : {np.sqrt(train_a_var)}")
print(f"Standard Variation of Train A is : {np.std(train_a)}")
```

```python
age = [28,26,21,19,32,42,52,12,16,78,12,34,57,85,23,56]
print(np.median(age))
print(np.percentile(age,50))
print(np.percentile(age,100))
print(np.percentile(age,0))
```

```python
q2 = np.percentile(age,50)
q1 = np.percentile(age,25)
q3 = np.percentile(age,75)
print(f"75th Percentile is : {q3}")
print(f"25th Percentile is : {q1}")
iqr = q3 - q1
print(iqr)
```

### Outlier :

* With the help of iqr, we find Outliers
* upper_limit = q3 + 1.5*iqr
* lower_limit = q1 - 1.5*iqr
* Any Data above upper limit and below lower limit is called as Outlier.

### Kernal Density Estimation (KDE)

* Approximate All these Bins
* PDF ---> Probability Density Function or Distributio Graph

```python
import seaborn as sns
height = [160,148,147,157,165,158,158,162,159,169,171,164,170,148,156,152]
sns.kdeplot(height)
```

```python
from scipy.stats import skew,kurtosis
import numpy as np
height = [160,157,165,158,158,162,159,169,164,148,156,152] # Remove Two Extreme Values from RHS and LHS
print(f"The Mean of data is : {np.mean(height)}")
print(f"The Skewness in the data is : {skew(height)}")
print(f"The Skewness in the data is : {kurtosis(height,fisher=False)}")
sns.kdeplot(height)
```

### Z Score :

* It is a value that represents a data point value with respect to it's standard deviation.
* Formula :
  - Z = (X - Mu)/Sigma

```python
temp = [30,35,23,26,21,39,33,27,25,30,32,35]
temp_mu = np.mean(temp)
temp_sigma = np.std(temp)
temp_normalized = [(i - temp_mu )/temp_sigma for i in temp]
print(temp_normalized)
```

```python
sns.kdeplot(temp)
```

```python
sns.kdeplot(temp_normalized)
```

```python
dist = [1200,1800,1900,1300,2000,2500,1900,2300,1000,2000,2500,1600]
dist_mu = np.mean(dist)
dist_sigma = np.std(dist)
dist_normalized = [(i-dist_mu)/dist_sigma for i in dist]
print(dist_normalized)
```

```python
sns.kdeplot(dist)
```

```python
sns.kdeplot(dist_normalized)
```
