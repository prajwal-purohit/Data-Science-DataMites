```python
age = [28,26,21,19,32,42,52,12,16,78,12,34,57,85,23,56]
print(np.median(age))
```

```python
q2 = np.percentile(age,50)
q1 = np.percentile(age,25)
q3 = np.percentile(age,75)
print("75th percentile", q3)
print("25th percentile", q1)
iqr = q3 - q1
print(iqr)
```

```python
# with the help of iqr, we find outliers
# upper_limit = q3 + 1.5*iqr
# lower_limit = q1 - 1.5*iqr
# Any data above upper limit and below lower limit 
# is called as outlier
```

```python
## Z score --> A value that represents a data point value with respect 
# to its standard deviation.
```

```python
temp = [30,35,23,26,21,39,33,27,25,30,32,35]
temp_mu = np.mean(temp)
temp_sigma = np.std(temp)
temp_normalized = []
for i in temp:
    temp_normalized.append((i - temp_mu)/temp_sigma)
print(temp_normalized)
print("The mean of temp_normalized is",np.mean(temp_normalized),"and std of tem_normalized is",np.std(temp_normalized))
```

```python
sns.kdeplot(temp)
```

```python
sns.kdeplot(temp_normalized)
```

```python
print(np.mean(temp_normalized))
print(np.std(temp_normalized))
```

```python
dist = [1200,1800,1900,1300,2000,2500,1900,2300,1000,2000,2500,1600]
dist_mu = np.mean(dist)
dist_sigma = np.std(dist)
print("mean of distance is",dist_mu,"and std dev of distance is",dist_sigma)
print("creating Z score for distance data")
dist_normalized = [(i-dist_mu)/dist_sigma for i in dist]
print(dist_normalized)
print("mean of distance_normalized is",np.mean(dist_normalized),"and std dev of distance_normalized is",np.std(dist_normalized))
```

```python
sns.kdeplot(dist)
```

```python
sns.kdeplot(dist_normalized)
```
